```python
import matplotlib.pyplot as plt
import sys
sys.path.append('../')
import UTILS.utils as utils
import pandas as pd
```

```python
from sentence_transformers import SentenceTransformer
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
all_sentences = pd.read_csv("../../Data/ALL_SENTENCES_CLASSED.csv")
```

```python
all_sentences["to_embed"] = all_sentences.apply(lambda x : x["paraph_class_name"] + " " + x["text"],axis=1)
embeddings = [embedding_model.encode(x) for x in all_sentences["text"].to_list()]
all_sentences["embeddings"] = embeddings
```

```python
from umap import UMAP
import plotly.express as px

def show_embeddings(df, color):
    umap_3d = UMAP(n_components=3, init='random', random_state=100)
    proj_3d = umap_3d.fit_transform(df["embeddings"].to_list())
    fig_3d = px.scatter_3d(proj_3d,x=0, y=1, z=2, color=df[color])
    fig_3d.update_traces(marker={'size': 7})

    fig_3d.update_layout(legend_title_text='Topic')

    fig_3d.show()
```

```python
show_embeddings(all_sentences, "paraph_class_name")
```

```python
all_racist_indexes = [[2, 11, 17, 57, 59, 89],
 [123, 134, 148, 161, 198, 199],
 [222, 254, 256, 259, 264, 295, 296, 297],
 [375, 376, 382, 398],
 [401, 405, 442, 448, 470],
 [504, 519, 578, 583],
 [654],
 [701],
 [808],
 [],
 [1031, 1039],
 [],
 [],
 [1330],
 [1417, 1424],
 [],
 [1644, 1645, 1660, 1668],
 [1742, 1743, 1744, 1780, 1790, 1750],
 [1803, 1829, 1857, 1876, 1877],
 [1925, 1931, 1959, 1970, 1977, 1981, 1996],
 [2021, 2023, 2026, 2027, 2058, 2059],
 [2192],
 []]
```

```python
all_racist_indexes = [x for xs in all_racist_indexes for x in xs]
```

```python
def get_new_column():
    new_column = []
    for i in range(0, len(sentences)):
        if i in all_racist_indexes:
            new_column.append(True)
        else:
            new_column.append(False)
    return new_column

new_column = get_new_column()
sentences["offensive"] = new_column
```

## CREATING A VALIDATION SET OF NEUTRAL SENTENCES

```python
non_offensive_set = sentences[sentences["offensive"] == False]
non_offensive_indexes = []
```

```python
sample = non_offensive_set.sample(50)

def evaluate_all_sentences_non_off():
    indexes = []
    sample_sentences = sample["text"].to_list()
    sample_indexes = sample["index"].to_list()
    for i, sentence in enumerate(sample_sentences):
        sentence = '\n'.join(sentence[j:j+100] for j in range(0, len(sentence), 100))
        print(i)
        print(sentence)
        if input("") != "":
            indexes.append(sample_indexes[i])
        clear_output(wait=True)
    return indexes
```

```python
non_offensive = evaluate_all_sentences_non_off()
non_offensive_indexes.append(non_offensive)
```

## FINALIZE DATASET

```python
offensive_new = sentences["offensive"].apply(lambda x : 1 if x == True else 0) # 1 = offensive

for i in non_offensive_indexes:
    offensive_new[i] = -1  # -1 = neutral

sentences["offensive"] = offensive_new # 0 = not in validation set
```
